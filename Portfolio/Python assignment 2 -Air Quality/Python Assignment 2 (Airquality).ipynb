## Task 3 (Missing values)

```python
# Count missing values per column
df.isna().sum()
```

```python
# Create a copy of the dataset
df_mv = df.copy()
```

```python
 # Fill missing values
# Fill Ozone with the column median
df_mv['Ozone'] = df_mv['Ozone'].fillna(df_mv['Ozone'].median())
```

```python
# Fill solar with the column mean
df_mv['Solar.R'] = df_mv['Solar.R'].fillna(df_mv['Solar.R'].mean())
```

```python
# Verify the new dataset has no missing values 
df_mv.isna().sum()
```

## Why is the median sometimes preferred over the mean

# The median is the middle position of any dataset after sorting, so the median is less affected by extremely high or low values. For this reason, the median is the better choice over the mean.

## Task 7 (Duplicates)

```python
# Check if there are any duplicate rows
df.duplicated().sum()
```

```python
# Create a few duplicate rows (First 3 rows)
df_dup = pd.concat([df, df.iloc[:3]], ignore_index=True)
print(df_dup.shape)
```

```python
# Check duplicates in the new DataFrame
df_dup.duplicated().sum()
```

```python
# Remove duplicates
df_clean=df_dup.drop_duplicates()
```

```python
# Report how many duplicates were 
removed = len(df_dup) - len(df_clean)
print("Duplicates removed:", removed)
```

## Task 8 (Derived columns)

```python
# Create Temp_C (Fahrenheit → Celsius)
df['Temp_C']=df['Temp']-32*5/9
```

```python
# Check the result
df[['Temp', 'Temp_C']].head()
```

```python
# Create a Boolean column High_Ozone
df['High_Ozone'] = df['Ozone'] > 100
```

```python
# Check the result
df[['Ozone', 'High_Ozone']].head()
```

```python
# Create a categorical column Temp_Band
df['Temp_Band'] = pd.cut(
    df['Temp'],
    bins=[0, 75, 85, float('inf')],
    labels=['Cool', 'Warm', 'Hot'],
    right=False
)
```

```python
# Check the result
df[['Temp', 'Temp_Band']].head(10)
```

## Task 9 (Groupby & aggregation)

```python
# Group by (Month)
df.groupby('Month').agg({
    'Ozone': 'mean',
    'Solar.R': 'median',
    'Temp': 'max'
})
```

```python
# Count how many days have Ozone > 80 in each month
df.groupby('Month')['Ozone'].apply(lambda x: (x > 80).sum())
```
