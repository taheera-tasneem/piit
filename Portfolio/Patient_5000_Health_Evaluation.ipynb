```python
from sqlalchemy import create_engine, text
import pymysql
import pandas as pd
import os
```

```python
engine1 = create_engine(
    "mysql+pymysql://root:piit2025@localhost:3306/patientcare"
)
print (engine1)
```

```python
# Function to run the query with parameters or query
def q(sql, **params):
    with engine1.connect() as conn:
        return pd.read_sql(text(sql), conn, params=params)
```

```python
q("SHOW TABLES")
```

```python
q("SELECT * FROM patients LIMIT 30")
```

```python
# Overall patient analysis
q("""SELECT  
    COUNT(*) AS total_patients,
    AVG(age) AS avg_age,
    MIN(AGE) AS younger_patient,
    MAX(AGE) AS oldest_patient
    FROM patients""")
```

```python
# Gender distribution
q("SELECT gender, COUNT(*) AS total_patients FROM patients GROUP BY gender")
```

```python
q = """
SELECT
    CASE
        WHEN age < 18 THEN 'Child'
        WHEN age BETWEEN 18 AND 39 THEN 'Young Adult'
        WHEN age BETWEEN 40 AND 59 THEN 'Middle Age'
        ELSE 'Senior'
    END AS age_group,
    COUNT(*) AS total_patients
FROM patients
GROUP BY age_group
ORDER BY total_patients DESC;
"""
pd.read_sql(q, engine1)
```

```python
# Upper
q("""
SELECT
    Patient_ID,
    UPPER(Disease) AS disease_upper
FROM patients
LIMIT 20;
""")
```

```python
# Upper
q("""
SELECT
    Patient_ID,
    UPPER(Disease) AS disease_upper
FROM patients
LIMIT 10;
""")
```

```python
# Lower
q("""
SELECT
    Patient_ID,
    LOWER(Outcome) AS outcome_lower
FROM patients
LIMIT 20;
""")
```

```python
# Concatenate
q("""
SELECT
    Patient_ID,
    CONCAT(Disease, ' - ', Outcome) AS patient_status
FROM patients
LIMIT 20;
""")
```

```python
# Length
q("""
SELECT
    Disease,
    LENGTH(Disease) AS disease_length
FROM patients
LIMIT 20;
""")
```

```python
# Left
q("""
SELECT
    Disease,
    LEFT(Disease, 5) AS first_five
FROM patients
LIMIT 20;
""")
```

```python
# Right
q("""
SELECT
    Disease,
    RIGHT(Disease, 6) AS last_six
FROM patients
LIMIT 20;
""")
```

```python
# All six string functions
q("""
SELECT
    Patient_ID,
    UPPER(Disease) AS disease_upper,
    LOWER(Outcome) AS outcome_lower,
    CONCAT(Disease, ' - ', Outcome) AS patient_status,
    LENGTH(Disease) AS disease_length,
    LEFT(Disease, 3) AS first_three,
    RIGHT(Disease, 3) AS last_three
FROM patients
LIMIT 20;
""")
```

```python
# Format (useful for cost)
q("""SELECT
    Treatment_Cost,
    FORMAT(Treatment_Cost, 2) AS formatted_cost
FROM patients;""")
```

```python
q("""SELECT
    Treatment_Cost,
    FORMAT(Treatment_Cost, 2) AS formatted_cost
FROM patients
LIMIT 10;""")
```

```python
# Concatenate with a separator
q("""SELECT
    Patient_ID,
    CONCAT_WS(' | ', Disease, Outcome, Smoking_Status)
    AS patient_summary
FROM patients
LIMIT 10;""")
```

```python
# LPAD (add character to the left)
q("""SELECT
    Patient_ID,
    LPAD(Age, 3, '0') AS age_code
FROM patients
LIMIT 5;""")
```

```python
# RPAD (add character to the right)
q("""SELECT
    RPAD(Disease, 15, '*') AS disease_format
FROM patients
LIMIT 5;""")
```
